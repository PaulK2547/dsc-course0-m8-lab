## Based on the assumption that missing injury values indicate that no injuries were reported, replace the missing values in the injury columns with zero

```python
df['Total.Fatal.Injuries']=df['Total.Fatal.Injuries'].fillna(0)
df['Total.Serious.Injuries']=df['Total.Serious.Injuries'].fillna(0)
df['Total.Uninjured']=df['Total.Uninjured'].fillna(0)
```

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

## Compute the total number of passengers. 
Assume that every occupant falls into exactly one category: fatal, serious, minor, or uninjured.

```python
df['Total.Passengers'] = (
    df['Total.Fatal.Injuries'] +
    df['Total.Serious.Injuries'] +
    df['Total.Minor.Injuries'] +
    df['Total.Uninjured']
)
```

# Compute the sum of serious and fatal injuries

```python
df['fatal.serious.injuries'] = (
    df['Total.Fatal.Injuries'] +
    df['Total.Serious.Injuries']
)
```

# Compute the proportion of passengers experiencing severe outcome

```python
df['fatal.seriuos.rate']=np.where(
    df['Total.Passengers']>0,
    df['fatal.serious.injuries']/df["Total.Passengers"], np.nan
)
```

**Aircraft.Damage**

# Replace missing values in the Aircraft.Damage column with zero.
The assumption is that Nan implies that the aircraft did not sustain any damages

```python
df['Aircraft.damage']=df['Aircraft.damage'].fillna(0)
```

# Create a column for Aircraft damage score 

```python
damage_map={
    'Destroyed': 3,
    'Substantial':2,
    'Minor':1,
    0:0

}
df['damage.score']=df['Aircraft.damage'].map(damage_map)
```
